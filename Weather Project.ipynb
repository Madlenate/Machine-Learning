```python
combine = pd.concat([test["target"], pd.Series(predictions, index = test.index)], axis = 1) #Combines both actual values and predicted values
combine.columns = ["actual", "predictions"]
```

```python
def create_predictions(predictors, weatther_main, reg):
    train = weather_main.loc[:"2020-12-31"]
    test = weather_main.loc["2021-01-01":]
    reg.fit(train[predictors], train["target"])
    predictions = reg.predict(test[predictors])
    error = mean_absolute_error(test["target"], predictions)
    combine = pd.concat([test["target"], pd.Series(predictions, index = test.index)], axis = 1)
    combine.columns = ["actual", "predictions"]
    return error, combine
    
```

```python
weather_main["month_max"] = weather_main["Max_Temp"].rolling(30).mean()
```

```python
weather_main["month_day_max"] = weather_main["month_max"] / weather_main["Max_Temp"]
```

```python
weather_main['max_min'] = weather_main['Max_Temp'] / weather_main['Min_Temp']
```

```python
predictors = ["Precipitation", "Max_Temp", "Min_Temp", "month_max", "month_day_max", "max_min"]
```

```python
weather_main = weather_main.iloc[30:,:].copy()
```

```python
error, combine = create_predictions(predictors, weather_main, reg)
```
